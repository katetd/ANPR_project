```python
import torch

print(torch.cuda.is_available())
print(torch.cuda.get_device_name(0))

!pip install -q kagglehub ultralytics

from pathlib import Path
import kagglehub

# Завантаження датасету
dataset_path = Path(
    kagglehub.dataset_download(
        "fareselmenshawii/license-plate-dataset"
    )
)

print(dataset_path)
```

```python
print(dataset_path)
```

```python
from pathlib import Path

for p in dataset_path.rglob("*"):
    if p.is_dir():
        print(p)
```

```python
yaml_text = f"""
path: {dataset_path}

train: images/train
val: images/val

names:
  0: license_plate
"""

with open("/content/dataset.yaml", "w") as f:
    f.write(yaml_text)
```

```python
from ultralytics import YOLO

model = YOLO("yolo11n.pt")

model.train(
    data="/content/dataset.yaml",
    epochs=30,
    imgsz=640,
    batch=16,
    workers=2,
    pretrained=True,
    project="/content/drive/MyDrive/OCR_Project/models",
    name="license_plate_detector"
)
```

```python
metrics = model.val()

print(metrics)
```

```python
results = model.predict(
    source=str(dataset_path / "images" / "val"),
    conf=0.25,
    save=True
)
```
